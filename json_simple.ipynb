# Json 

### 1. 1階層のJson

**シンプルに1階層のJsonを作る**


```python
import json
gunma = {'群馬県 前橋市':332149,'群馬県 高崎市':372973,'群馬県 沼田市':45337}
print(gunma)
```

**高崎市の人口を取り出せ**

```python
gunma['群馬県 高崎市']
```

### 2. 2階層のJson 
**県を第1階層、市区を第2階層とするJson**

```python
tg = {'群馬県':{'前橋市':332149, '高崎市':372973,'沼田市':45337},
       '東京都':{'千代田区':66680, '中央区':169179, '世田谷区':943664} }
print(tg)
```

**高崎市の人口を取り出せ**

```python
print(tg['群馬県']['高崎市'])
```
